```python
#第一次作业第二题
'''
二次方程的两个根
'''
a = int(input('请输入a'))
b = int(input('请输入b'))
c = int(input('请输入c'))
delta1 = pow(b,2) - 4*a*c
answer1 = (-b+delta1)/2*a
answer2 = (-b-delta1)/2*a
if(delta1 < 0):
  print("该方程无解")
else:
  print(answer1)
  print(answer2)
```
