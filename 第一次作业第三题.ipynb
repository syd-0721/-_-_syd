```python
#第一次作业第三题
'''
商店找钱
'''
a = int(input("请输入应该找的整数钱:"))
if(a>=100):
  print('错误了，请你输入小于100块钱的')
else:
  fifty = a//50
  a %= 50
  five = a//5
  a %= 5
  one = a//1
print("应该找%d张50元的，%d张5元的，%d张1元的"%(fifty,five,one))
```
